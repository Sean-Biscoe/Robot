# 01 · Video → Frames

Extracts every 5th frame from each raw lap recording (`Videos/*.mp4`) into
`Frames/<video_name>/frame_00000.jpg, ...`. Sampling every 5th frame removes
near-duplicate frames while keeping the dataset a manageable size
(5,632 frames total across 12 laps, 6 per direction).

Requires `opencv-python` (`pip install -r requirements.txt`).

```python
import cv2
import os

video_folder = "Videos"
frame_folder = "Frames"

os.makedirs(frame_folder, exist_ok=True)

frame_skip = 5

for video in os.listdir(video_folder):

    video_path = os.path.join(video_folder, video)

    # Skip non-video files
    if not video.endswith((".mp4", ".avi", ".mov", ".mkv")):
        continue

    # Create output folder for this video
    video_name = os.path.splitext(video)[0]
    save_folder = os.path.join(frame_folder, video_name)
    os.makedirs(save_folder, exist_ok=True)

    cap = cv2.VideoCapture(video_path)

    frame_id = 0
    saved_count = 0

    while True:
        ret, frame = cap.read()

        if not ret:
            break

        if frame_id % frame_skip == 0:
            filename = f"frame_{saved_count:05d}.jpg"
            cv2.imwrite(os.path.join(save_folder, filename), frame)
            saved_count += 1

        frame_id += 1

    cap.release()

    print(f"{video_name}: {saved_count} frames saved")

print("Done.")
```
